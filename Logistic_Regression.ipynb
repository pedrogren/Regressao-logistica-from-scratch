```python
def predict(self, x): # predição do usuario
        probabilidades = []
        for indice, linha in x.iterrows():
            somatorio = np.dot(self.w.T, linha.values) + self.b
            sigmoid = 1 / (1 + np.exp(-somatorio))
            probabilidades.append(sigmoid[0])
        return (np.array(probabilidades) >= 0.5).astype(int)
```
